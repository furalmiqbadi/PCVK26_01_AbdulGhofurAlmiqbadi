Install library Numpy, panda, cv2, PIL, matplotlob

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow #display image
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt
```

Input image dan menampilkannya setelah disesuaikan

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://i.pinimg.com/736x/dd/51/f3/dd51f327bb4a3903ef2ac8a6d16283d7.jpg"]

for url in urls:
  image = io.imread(url) #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert to RGB
  final_frame = cv.hconcat((image, image_2)) # horizontalconcatenate image
  cv2_imshow(final_frame) #show image
  print('\n')
```

Melihat Ukuran Image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("Resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

Membuat garis diagonal

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
